```python
import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
```

```python
df_week1 = pd.read_csv('XKO_HPD_week_1.csv')
df_week1.sort_values(by=['traceid','time'])
df_week1.head()
```

```python
df_week1.describe()
```

```python
df_week1.shape
```

```python
df_week1['traceid'].nunique()
```

```python
df_week1 = df_week1.groupby("traceid").filter(lambda x: len(x) >= 5)
```

```python
min_lon, min_lat, max_lon, max_lat = [21.088588, 42.571255, 21.188588, 42.671255]

df_week1 = df_week1[
    (df_week1["longitude"] >= min_lon) &
    (df_week1["longitude"] <= max_lon) &
    (df_week1["latitude"]  >= min_lat) &
    (df_week1["latitude"]  <= max_lat)
]

print(df_week1.shape)
```

```python
df_week1.info()
```

```python
df_week1["time"] = pd.to_datetime(df_week1["time"])
```

```python
df_week1.info()
```

```python
df_week1 = df_week1.sort_values(["traceid", "time"])

df_week1["order"] = df_week1.groupby("traceid").cumcount()

df_week1["trip_size"] = df_week1.groupby("traceid")["traceid"].transform("size")

df_week1["rel_pos"] = df_week1["order"] / (df_week1["trip_size"] - 1)

df_week1["is_null_head"] = df_week1["heading"].isna()
```

# missing data analysis

```python
missing_report = pd.DataFrame({
    "column": df_week1.columns,
    "null_count": df_week1.isna().sum().values,
    "null_pct": (df_week1.isna().mean().values * 100).round(4)
}).sort_values("null_pct", ascending=False)
missing_report
```

# head

```python
nulls = df_week1[df_week1["heading"].isna()].copy()
```

```python
nulls.head()
```

```python
def zone(x):
    if x <= 0.33:
        return "start"
    elif x <= 0.66:
        return "middle"
    else:
        return "end"

nulls["trip_zone"] = nulls["rel_pos"].apply(zone)
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=nulls, x="trip_zone", order=["start", "middle", "end"])

plt.xlabel("Trip position")
plt.ylabel("Count of NULL head")
plt.title("Where NULL headings occur")
plt.show()
```

```python
zone_percent_dict = (
    nulls["trip_zone"]
    .value_counts(normalize=True)
    .reindex(["start", "middle", "end"], fill_value=0) * 100
).round(2).to_dict()

print(zone_percent_dict)
```

```python
df_week1.groupby("is_null_head")["speed"].describe()
```

# Insight on Missing Headings

- **Start and End of Trips**
  - The majority of `NULL` values occur at the **start and end of trips**.  
  - Investigation revealed that the **sensors do not reliably detect headings** at the beginning and end of trips.

- **Low Speeds**
  - A significant portion of `NULL` values appears at **low speeds**.  
  - This is consistent with the same sensor limitation: heading detection is **less reliable when the vehicle is moving slowly**.
